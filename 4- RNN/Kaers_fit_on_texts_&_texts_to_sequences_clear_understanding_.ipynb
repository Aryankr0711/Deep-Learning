```python
import tensorflow
from tensorflow.keras.preprocessing.text import Tokenizer
```

```python
data="Hii My name is Aryan Kumar, currently i am a second year student"
```

```python
tokenizer=Tokenizer()
```

```python
tokens=tokenizer.fit_on_texts([data])
```

```python
print(tokenizer.word_index)
```

```python
sequences = tokenizer.texts_to_sequences([data])
print(sequences)
```
