```python
import tensorflow
from tensorflow import keras
from keras.layers import Dense, Conv2D, Flatten, MaxPooling2D, Dropout
from keras.models import Sequential
```

```python
model=Sequential()

model.add(Conv2D(96, kernel_size=(11, 11), strides=(4, 4), activation='relu',padding='valid', input_shape=(227, 227, 3)))
model.add(MaxPooling2D(pool_size=(3, 3), strides=(2, 2)))

model.add(Conv2D(256, kernel_size=(5, 5), strides=(1, 1), activation='relu', padding='same'))
model.add(MaxPooling2D(pool_size=(3, 3), strides=(2, 2)))

model.add(Conv2D(384, kernel_size=(3, 3), strides=(1, 1), activation='relu', padding='same'))
model.add(Conv2D(384, kernel_size=(3, 3), strides=(1, 1), activation='relu', padding='same'))
model.add(Conv2D(256, kernel_size=(3, 3), strides=(1, 1), activation='relu', padding='same'))
model.add(MaxPooling2D(pool_size=(3, 3), strides=(2, 2)))

model.add(Flatten())
model.add(Dense(4096, activation='relu'))
model.add(Dropout(0.5))
model.add(Dense(4096, activation='relu'))
model.add(Dropout(0.5))
model.add(Dense(10, activation='softmax')) # Changed to 10 classes for CIFAR-10
```

```python
model.summary()
```

# using Pre-trained model for classification

```python
from tensorflow.keras.applications.resnet50 import ResNet50, preprocess_input, decode_predictions
from tensorflow.keras.preprocessing import image
import numpy as np
```

```python
model2=ResNet50(weights='imagenet')
```

```python
model2.summary()
```

# Apron

```python
img_path_1='/content/apron.jpg'
img1=image.load_img(img_path_1, target_size=(224, 224))
x=image.img_to_array(img1)
x=np.expand_dims(x, axis=0)
x=preprocess_input(x)
```

```python
preds=model2.predict(x)
print('Predicted: ', decode_predictions(preds, top=3)[0])
```
