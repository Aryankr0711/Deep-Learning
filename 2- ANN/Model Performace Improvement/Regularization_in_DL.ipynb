```python
import tensorflow
from tensorflow import keras
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense
from sklearn.datasets import make_moons
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
X, y=make_moons(100, random_state=2, noise=0.2)
```

```python
sns.scatterplot(x=X[:, 0], y=X[:, 1], hue=y)
plt.show()
```

```python
model=Sequential()
```

```python
model.add(Dense(128, activation='relu', input_dim=2))
model.add(Dense(1, activation='sigmoid'))
```

```python
model.compile(loss='binary_crossentropy', optimizer='Adam')
```

```python
history=model.fit(X, y, validation_split=0.2, epochs=2000)
```

```python
plt.plot(history.history['loss'], label='Train')
plt.plot(history.history['val_loss'], label='val')
plt.legend()
plt.show()
```

```python
model=Sequential()
```

```python
model.add(Dense(128, activation='relu', input_dim=2, kernel_regularizer=tensorflow.keras.regularizers.l2(0.02)))
model.add(Dense(1, activation='sigmoid'))
```

```python
model.compile(loss='binary_crossentropy', optimizer='Adam')
```

```python
history=model.fit(X, y, validation_split=0.2, epochs=2000)
```

```python
plt.plot(history.history['loss'], label='Train')
plt.plot(history.history['val_loss'], label='val')
plt.legend()
plt.show()
```
